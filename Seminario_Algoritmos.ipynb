Sea X una expresión compuesta por 5 cifras y 4 signos, alternados de forma que cumplen f(X) - n = 0, para cada n provisto (f puede pensarse como el eval).

El problema planteado puede verse como un problema de minimización si queremos encontrar una combinación que cumpla la condición antes mencionada (f(X) - n = 0). En la práctica esto no resulta tan así dado que el tamaño limitado del problema permite atacarlo de forma exacta.

Diseña un algoritmo para resolver el problema por fuerza bruta

```python
import itertools

numeros = [x for x in range(1, 10)]
signos = ["+","-","*","/"]

def fuerza_bruta(objetivo):
    for permutaciones_numeros in itertools.permutations(numeros, 5):
        values = permutaciones_numeros

        for permutaciones_signos in itertools.permutations(signos):
            operadores = permutaciones_signos
            expresion = f"{values[0]}{operadores[0]}{values[1]}{operadores[1]}{values[2]}{operadores[2]}{values[3]}{operadores[3]}{values[4]}"
            resultado = eval(expresion)

            if resultado == objetivo:
                return expresion
    return None
```

```python
# Algunos ejemplos de prueba

import numpy as np

valores_a_calcular = np.random.randint(-100, 100, 20)

for valor in valores_a_calcular:
    expression = fuerza_bruta(valor)

    if expression:
        print(f"{valor} = {expression}")
    else:
        print(f"{valor} no puede ser generado con las restricciones dadas.")
```

Para mejorar el resultado de la solución por fuerza bruta, se realizó un análisis orientado desde el lado matemático.

En primer lugar, se puede ver que los operadores disponibles tienen cierto efecto amplificador o reductor de las cardinalidades, en particular se rigen por el siguiente "orden":
- Multiplicación (*): operador que más permite amplificar. 9*8 es ~4 veces mayor que 9+8.
- Suma (+): amplifica en menor medida, pero añade valor positivo.
- División: reduce el valor original, pero tiene un efecto moderado. En particular el resultado estará entre 2 y n / 2.
- Resta: reduce en mayor medida ya que es una substracción directa.

Teniendo en cuenta esto, se procedió a calcular los valores máximos y mínimos posibles:
- **Máximo: 9*8+7-2/1 = 77**. Se puede verificar que no hay ninguna cifra que se pueda reemplazar para aumentar este valor.
- **Mínimo: 1-9*8+4/2 = -69**. De igual manera en este caso, no se puede reemplazar ninguna cifra para reducir este valor.

Como siguiente paso y teniendo en cuenta los límites de los valores que se pueden generar, se procedió a realizar algunas pruebas manuales que llevaron a la conclusión de que generalizando la "forma" de las expresiones máximo y mínimo anteriores, se pueden generar todos los valores posibles entre ellos.

Dichas fórmulas son representadas de la siguiente manera:
- Si el objetivo (n) es mayor que cero: a * b + c - d / e
- Si n es menor o igual que cero: c - a * b + d / e

La sola implementación de esta fórmula elimina la necesidad de permutar los operadores, lo que efectivamente reduce el número de posibilidades a 9! / 4! = 15120 lo cual es efectivamente 1/24 veces la cantidad de posibilidades a evaluar por fuerza bruta.

Como siguiente paso, se puede ver que ambas fórmulas poseen dos partes: las cifras multiplicadas y las cifras que suman o restan. Un breve análisis muestra que la dinámica general es: elegir n1*n2 para caer aproximadamente cerca del objetivo y luego definir las cifras restantes en base a la diferencia existente.

Para optimizar este paso, se procedió al precálculo de n1*n2, lo cual genera una matriz de aproximadamente n^2 y con una complejidad a su vez de n^2.

```python
multiplos = {x: [] for x in range(0, 8)}

for index, i in enumerate(numeros):
    for j in numeros[index + 1:]:
        decena = (i * j) // 10
        multiplos[decena].append((i,j))

for key, value in multiplos.items():
    print(f"{key}: {value}")
```

Explicación del algoritmo anterior:
- Se precalculan todas las combinaciones posibles de i * j.
- Como hemos visto antes, el máximo i * j es 9 * 8 = 72, con lo que es suficiente para nuestro diccionario de múltiplos tener las claves de 0 a 7 inclusive.
- Para j se toma siempre un valor mayor que i. Así evitamos i*i y también terminos conmutados (i * j y j * i son iguales)
- El algoritmo se ejecuta en un múltiplo (menor) de n^2.

Un detalle interesante que surje del precálculo de los múltiplos por decena es que, como aplicaremos más adelante, se reduce bastante el número de posibilidades a evaluar. Originalmente para 9 cifras distintas, podríamos elegir entre 9 opciones para el primero y 8 opciones para el segundo. A esto hay que dividir por 2, dado que la multiplicación es conmutativa y nos quedan 36 posibilidades.
Podemos ver en la lista anterior, que al elegir el par de números multiplicados a partir de la decena en la que cae la multiplicación, podemos reducir considerablemente las posibilidades, a un máximo de 10 en el peor de los casos (para los números de 0 a 9).

Por otra parte, todavía tenemos que calcular los valores restantes. Recordemos:
- c - d / e si objetivo > 0
- c + d / e si objetivo <= 0

Aquí podemos realizar más optimizaciones.

En primer lugar, podríamos considerar que las posibles combinaciones de cifras son 7 * 6 * 5 = 210. Pero al analizar un poco más en profundidad podemos ver que:
- c está libre: 7 posibilidades.
- e debe ser < 4 y < d. Màximo 4 posibilidades.
- d debe ser > e. En principio 5 posibilidades.

Hemos reducido el espacio de combinaciones a 7 * 5 * 4 = 140. No parece una mejora tan representativa, pero podemos aplicar un precàlculo interesante y poco costoso: calcular què nùmeros son divisibles por cada número disponible.

```python
divisores = {x: [] for x in range(1,5)}

for index, i in enumerate(numeros[0:4]):
    for j in numeros[index + 1:]:
        if j % i == 0:
            divisores[i].append(j)

for key, value in divisores.items():
    print(f"{key}: {value}")
```

Aplicando una lógica sobre los posibles valores de d dependientes del e que hemos elegido, podemos ver que efectivamente las combinaciones posibles se reducen aún más. La cuenta ahora es un poco más complicada, pero podemos intentar aproximarla de la siguiente manera:
- c tendrá 7 posibilidades.
- e tendrá 4 posibilidades: 1,2,3,4
- d tendrá las siguientes posibilidades:
    - Si e vale 1, quedarán 5 valores disponibles luego de fijar c y d.
    - Si e vale 2, 3 posibilidades.
    - Si e vale 3, 2 posibilidades.
    - Si e vale 4, 1 única posibilidad.

Uniendo todo esto vemos que por cada c consideramos 11 posibilidades, con lo cual se reducen las posibilidades a 7 * 11 = 77.

A esta altura podemos construir la primera mitad del algoritmo: 0 < objetivo <= 77 (màximo).

```python
def encontrar_sumandos(numeros_para_c, diferencia):
    for c in numeros_para_c:
        resultado_division = c - diferencia

        numeros_para_e = list(set(numeros_para_c).difference([c, 5,6,7,8,9]))
        for e in numeros_para_e:
            numeros_para_d = list(set(numeros_para_c).difference({e}).intersection(divisores[e]))
            if (resultado_division * e) in numeros_para_d:
                return c, (resultado_division * e), e

    return None

def encontrar_formula_n_positivo(n):
    decena = n // 10
    for t in multiplos[decena]:
        diferencia = n - t[0] * t[1]
        numeros_para_c = list(set(numeros).difference(t))

        res = encontrar_sumandos(numeros_para_c, diferencia)

        if res is not None:
            c, d, e = res
            return t[0], t[1], c, d, e
    return None
```

```python
for i in range(1, 78):
    res = encontrar_formula_n_positivo(i)
    if res is not None:
        a,b,c,d,e = res
        print(f"{i} = {a}*{b}+{c}-{d}/{e}")
    else:
        print(f"No se encontró fórmula para {i}")
```

Para los valores -69 (mínimo) <= objetivo <= 0 podemos realizar un procedimiento similar, pero utilizando la otra fórmula descubierta.

Un par de detalles a tener en cuenta:
- El único término negativo será el multiplicativo, con lo que los otros dos términos estarán sumando y disminuyendo el valor absoluto. Esto lleva a la necesidad de comenzar la búsqueda con un a * b menor al objetivo. Por ejemplo, si queremos obtener -64, no podemos utilizar 9 * 7 = 63 porque no tenemos más términos negativos para llegar a -64 (la fórmula sería c - 9 * 7 + d / e y sólo puede dar valores > -63)
- Dada esta condición, no nos alcanza con verificar solamente las multiplicaciones que caen en la misma decena del número objetivo. Debemos también verificar la decena + 1.

```python
def encontrar_sumandos_positivos(numeros_para_c, diferencia):
    for c in numeros_para_c:
        resultado_division = diferencia - c

        numeros_para_e = list(set(numeros_para_c).difference([c, 5,6,7,8,9]))
        for e in numeros_para_e:
            numeros_para_d = list(set(numeros_para_c).difference({c,e}).intersection(divisores[e]))
            if (resultado_division * e) in numeros_para_d:
                return c, (resultado_division * e), e

    return None

def encontrar_formula_n_negativo(n):
    decena = (n * (-1)) // 10

    multiplos_a_validar = [t for t in multiplos[decena] + multiplos[decena + 1] if t[0] * t[1] * (-1) <= n]

    for t in multiplos_a_validar:
        diferencia = n + t[0] * t[1]
        numeros_para_c = list(set(numeros).difference(t))

        res = encontrar_sumandos_positivos(numeros_para_c, diferencia)

        if res is not None:
            c, d, e = res
            return t[0], t[1], c, d, e
    return None
```

```python
for i in range(-69, 1):
    res = encontrar_formula_n_negativo(i)
    if res is not None:
        a,b,c,d,e = res
        print(f"{i} = {c}-{a}*{b}+{d}/{e}")
    else:
        print(f"No se encontró fórmula para {i}")
```
